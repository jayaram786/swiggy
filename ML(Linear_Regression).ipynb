```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(r"C:\Users\HP\Downloads\ML practice\homeprices.csv")
df
```

```python
%matplotlib inline
plt.xlabel('Area(sqrt ft)')
plt.ylabel('price(US$)')
plt.scatter(df.area,df.price,color = 'purple',marker ='*')
plt.show()
```

```python
new_df = df.drop('price',axis='columns')
new_df
```

```python
price = df.price
price
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

# to predict home price with area = 3300 sqrt ft

```python
reg.predict([[3300]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
# y = mx+b
135.78767123*3300+180616.43835616432
```

# reg.predict &  y=mx+b get same values

```python
# representation of linear regression
%matplotlib inline
plt.xlabel('Area', fontsize = 20)
plt.ylabel('price',fontsize = 20)
plt.scatter(df.area,df.price,color = 'red',marker ='*')
plt.plot(df.area,reg.predict(df[['area']]), color = 'blue')
plt.show()
```
